```python
import numpy as np
```

```python
# Define matrix A
A = np.array([[2, 0], [0, 5]])
```

```python
# Objective function
def f(x):
    return np.dot(x.T, np.dot(A, x))

# Gradient
def grad_f(x):
    return 2 * np.dot(A, x)

# Constant step size
def gd_constant_alpha(x0, alpha, num_iterations):
    x = x0.copy()
    for _ in range(num_iterations):
        x = x - alpha * grad_f(x)
    return x, f(x)

# Exact line search
def gd_exact_line_search(x0, num_iterations):
    x = x0.copy()
    for _ in range(num_iterations):
        grad = grad_f(x)
        alpha = np.dot(grad, grad) / (2 * np.dot(grad, np.dot(A, grad)))
        x = x - alpha * grad
    return x, f(x)

# Backtracking line search
def backtracking_line_search(x, grad, gamma=0.5, sigma=0.2):
    alpha = 1.0
    f_x = f(x)
    grad_norm_sq = np.dot(grad, grad)
    while f(x - alpha * grad) > f_x - gamma * alpha * grad_norm_sq:
        alpha *= sigma
    return alpha

def gd_backtracking(x0, num_iterations, gamma=0.5, sigma=0.2):
    x = x0.copy()
    for _ in range(num_iterations):
        grad = grad_f(x)
        alpha = backtracking_line_search(x, grad, gamma, sigma)
        x = x - alpha * grad
    return x, f(x)
```

```python
# Initial point
x0 = np.array([1.0, 1.0])

# Run for 10 iterations
x_const, f_const = gd_constant_alpha(x0, 0.1, 10)
x_exact, f_exact = gd_exact_line_search(x0, 10)
x_back, f_back = gd_backtracking(x0, 10)
```

```python
# Results
print("Constant step size (alpha = 0.1):")
print(f"Last iterate: {x_const}")
print(f"Objective value: {f_const:.10f}\n")

print("Exact line search:")
print(f"Last iterate: {x_exact}")
print(f"Objective value: {f_exact:.10f}\n")

print("Backtracking line search:")
print(f"Last iterate: {x_back}")
print(f"Objective value: {f_back:.10f}")
```
