```python
import pandas as pd
import numpy as np
from scipy import stats
import statsmodels.api as sm
```

# g) Fit the model $y_i \stackrel{i.i.d}{\sim} N(\mu, \sigma^2)$ to the data

```python
# File path
file_path = r"D:\MG\！CUHKSZ\Berkeley 交流\❗课\STAT 153 - Intro to Time Series\Homework\1_CPIAUCSL.xlsx"

# Read data from Excel file
df = pd.read_excel(file_path, sheet_name='Monthly')
print(f"Total rows (including header): {len(df) + 1}")

# Keep data from January 1947 to July 2025 (exclude August 2025)
df = df.head(944)  # Keep first 943 rows (including header)

# Extract CPI values
cpi = df['CPIAUCSL'].values

# Calculate log differences (monthly inflation rates)
log_cpi = np.log(cpi)
y = np.diff(log_cpi)
n = len(y)

print(f"Number of observations: {n}")
```

# h) Estimate and Uncertainty Interval for $\mu$

```python
# Calculate estimate and uncertainty interval for μ
mean_y = np.mean(y)
std_y = np.std(y, ddof=1)
se_mean = std_y / np.sqrt(n)
t_val = stats.t.ppf(0.975, n-1)
ci_low = mean_y - t_val * se_mean
ci_high = mean_y + t_val * se_mean
```

```python
print(f"Estimate for μ: {mean_y}")
print(f"95% uncertainty interval for μ: ({ci_low}, {ci_high})")
```

# i) Comparison with Linear Regression Model

```python
# Compare with linear regression model
indices = np.arange(1, len(log_cpi))
X = sm.add_constant(indices)
model = sm.OLS(log_cpi[1:], X)  # Use same observations as y
results = model.fit()
beta1 = results.params[1]
se_beta1 = results.bse[1]
t_val_reg = stats.t.ppf(0.975, results.df_resid)
ci_beta1_low = beta1 - t_val_reg * se_beta1
ci_beta1_high = beta1 + t_val_reg * se_beta1
```

```python
print(f"\nLinear regression model results:")
print(f"Estimate for β1: {beta1}")
print(f"95% confidence interval for β1: ({ci_beta1_low}, {ci_beta1_high})")
```

```python
# Compare the two intervals
print(f"\nComparison:")
print(f"Width of μ's interval: {ci_high - ci_low}")
print(f"Width of β1's interval: {ci_beta1_high - ci_beta1_low}")
```
