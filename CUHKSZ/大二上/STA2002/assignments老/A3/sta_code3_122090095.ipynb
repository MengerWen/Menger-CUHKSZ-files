```python
#q4(b)
import pandas as pd

# Load the dataset
data = pd.read_csv('AB_2020.csv')

# Filter the dataset to select only the 'landing page' and 'converted' columns
selected_data = data[['landing_page', 'converted']]

# Count the number of conversions and total users for each landing page (A and B)
conversion_counts = selected_data.groupby('landing_page')['converted'].sum()
total_users = selected_data['landing_page'].value_counts()

# Calculate the maximum likelihood estimates for pA and pB
pA = conversion_counts['A'] / total_users['A']
pB = conversion_counts['B'] / total_users['B']

print(f"Maximum Likelihood Estimate of pA: {pA:.4f}")
print(f"Maximum Likelihood Estimate of pB: {pB:.4f}")
```

```python
#q4(c)
import pandas as pd
import numpy as np
import scipy.stats as stats

# Load the dataset
data = pd.read_csv('AB_2020.csv')

# Select data for Version A and Version B
data_A = data[data['landing_page'] == 'A']
data_B = data[data['landing_page'] == 'B']

# Calculate the proportions of users who clicked in Version A and Version B
pA = np.mean(data_A['converted'])
pB = np.mean(data_B['converted'])

# Calculate the standard error for the difference in proportions
nA = len(data_A)
nB = len(data_B)
se = np.sqrt(pA * (1 - pA) / nA + pB * (1 - pB) / nB)

# Calculate the test statistic (Z-score)
Z = (pA - pB) / se

# Calculate the two-tailed p-value
p_value = 2 * (1 - stats.norm.cdf(np.abs(Z)))

print(f"Test Statistic (Z): {Z:.4f}")
print(f"P-value: {p_value:.4f}")
```

```python
# q5
import pandas as pd
import numpy as np
from scipy.stats import t

# 从CSV文件加载数据
data = pd.read_csv("traffic.csv")

# 从数据中提取 "weather_main" 和 "traffic_volume" 列
clear_traffic = data[data['weather_main'] == "Clear"]["traffic_volume"]
rain_traffic = data[data['weather_main'] == "Rain"]["traffic_volume"]

# 计算均值和标准差
x_mean = np.mean(clear_traffic)
y_mean = np.mean(rain_traffic)
x_std = np.std(clear_traffic, ddof=1)  # 使用 ddof=1 进行自由度校正
y_std = np.std(rain_traffic, ddof=1)

# 样本大小
n = len(clear_traffic)
m = len(rain_traffic)

# (b) 假设等方差
# 计算 t 统计量
t_statistic = (x_mean - y_mean) / np.sqrt((x_std**2 / n) + (y_std**2 / m))

# 计算 t 临界值
alpha = 0.05
t_critical = t.ppf(1 - alpha, df=n + m - 2)

# 进行假设检验
if t_statistic < -t_critical:
    print("(b) t_statistic is:"+str(t_statistic)+" t_critical is:"+str(t_critical))
    print("So reject H0: μX > μY")
else:
    print("(b) Fail to reject H0")

# (c) 假设非等方差
# 使用 Welch's t-test
t_statistic_welch = (x_mean - y_mean) / np.sqrt((x_std**2 / n) + (y_std**2 / m))

# 计算 t 临界值（自由度使用 Welch-Satterthwaite 公式）
df_welch = ((x_std**2 / n + y_std**2 / m)**2) / ((x_std**4 / (n**2 * (n - 1))) + (y_std**4 / (m**2 * (m - 1))))
t_critical_welch = t.ppf(1 - alpha, df=df_welch)

# 进行假设检验
if t_statistic_welch < -t_critical_welch:
    print("(c) t_statistic_welch is:"+str(t_statistic_welch)+" t_cirtical_welch is:"+str(t_critical_welch))
    print(" So reject H0: μX > μY")
else:
    print("(c) Fail to reject H0")

# (d) 进行 F-Test
# H0: σ2X = σ2Y, H1: σ2X ≠ σ2Y
f_statistic = (x_std**2) / (y_std**2)

# 计算 F 临界值
f_critical_low = stats.f.ppf(alpha / 2, dfn=n - 1, dfd=m - 1)
f_critical_high = stats.f.ppf(1 - alpha / 2, dfn=n - 1, dfd=m - 1)

# 进行假设检验
if f_statistic < f_critical_low or f_statistic > f_critical_high:
    print("(d) Reject H0: σ2X ≠ σ2Y")
else:
    print("(d) f_statistic is:"+str(f_statistic)+" f_critical_low is:"+ str(f_critical_low)+" f_critical_high is:"+str(f_critical_high))
    print("So fail to reject H0")
```
