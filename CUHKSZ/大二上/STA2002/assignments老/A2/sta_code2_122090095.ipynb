```python
# Q6(b)(c)
import pandas as pd
import numpy as np
from scipy import stats

# 从CSV文件加载数据
data = pd.read_csv("traffic.csv")

# 从数据中提取 "weather_main" 和 "traffic_volume" 列
clear_traffic = data[data['weather_main'] == "Clear"]["traffic_volume"]
rain_traffic = data[data['weather_main'] == "Rain"]["traffic_volume"]

# 计算均值和标准差
x_mean = np.mean(clear_traffic)
y_mean = np.mean(rain_traffic)
x_std = np.std(clear_traffic, ddof=1)  # 使用 ddof=1 进行自由度校正
y_std = np.std(rain_traffic, ddof=1)

# 样本大小
n = len(clear_traffic)
m = len(rain_traffic)

# (b) 假设等方差
# 计算合并标准差（pooled standard deviation）
pooled_std = np.sqrt(((n - 1) * x_std**2 + (m - 1) * y_std**2) / (n + m - 2))

# 计算 z 临界值
alpha = 0.05
z_critical = stats.norm.ppf(1 - alpha/2)

# 计算标准误差
standard_error = pooled_std * np.sqrt(1/n + 1/m)

# 计算置信区间
margin_of_error = z_critical * standard_error
confidence_interval = (x_mean - y_mean - margin_of_error, x_mean - y_mean + margin_of_error)

# 打印结果
print(f"(b) 均值差: {x_mean - y_mean}")
print(f"(b) 95% 置信区间: {confidence_interval}")

```

```python
import pandas as pd
import numpy as np
from scipy.stats import t

# Load the dataset
data = pd.read_csv("traffic.csv")

# Filter data for "Clear" and "Rain" weather conditions
clear_traffic = data[data['weather_main'] == "Clear"]
rain_traffic = data[data['weather_main'] == "Rain"]

# Calculate sample means and sample standard deviations
x_mean = clear_traffic['traffic_volume'].mean()
y_mean = rain_traffic['traffic_volume'].mean()
x_std = clear_traffic['traffic_volume'].std()
y_std = rain_traffic['traffic_volume'].std()

# Sample sizes
n = len(clear_traffic)
m = len(rain_traffic)

# Degrees of freedom
df_x = n - 1
df_y = m - 1

# Calculate the critical value for a two-tailed test with alpha = 0.05/2 = 0.025
alpha = 0.025
z_critical = t.ppf(1 - alpha, df_x + df_y)

# Calculate the margin of error
margin_error = z_critical * np.sqrt((x_std**2 / n) + (y_std**2 / m))

# Calculate the confidence interval
lower_bound = (x_mean - y_mean) - margin_error
upper_bound = (x_mean - y_mean) + margin_error

print("95% Welch's t-Interval for the difference μX - μY:")
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
# Q7
import pandas as pd
import numpy as np
from scipy.stats import t
from scipy import stats


# 从CSV文件加载数据
data = pd.read_csv("traffic.csv")

# 从数据中提取 "weather_main" 和 "traffic_volume" 列
clear_traffic = data[data['weather_main'] == "Clear"]["traffic_volume"]
rain_traffic = data[data['weather_main'] == "Rain"]["traffic_volume"]

# 计算均值和标准差
x_mean = np.mean(clear_traffic)
y_mean = np.mean(rain_traffic)
x_std = np.std(clear_traffic, ddof=1)  # 使用 ddof=1 进行自由度校正
y_std = np.std(rain_traffic, ddof=1)

# 样本大小
n = len(clear_traffic)
m = len(rain_traffic)

# (b) 假设等方差
# 计算 t 统计量
t_statistic = (x_mean - y_mean) / np.sqrt((x_std**2 / n) + (y_std**2 / m))

# 计算 t 临界值
alpha = 0.05
t_critical = t.ppf(1 - alpha, df=n + m - 2)

# 进行假设检验
if t_statistic < -t_critical:
    print("(b) t_statistic is:" + str(t_statistic) + "  t_critical is:" + str(-t_critical))
    print(" So reject H0: μX > μY")
else:
    print("(b) t_statistic is:" + str(t_statistic) + "t_critical is:" + t_critical)
    print(" Fail to reject H0")

# (c) 假设非等方差
# 使用 Welch's t-test
t_statistic_welch = (x_mean - y_mean) / np.sqrt((x_std**2 / n) + (y_std**2 / m))

# 计算 t 临界值（自由度使用 Welch-Satterthwaite 公式）
df_welch = ((x_std**2 / n + y_std**2 / m)**2) / ((x_std**4 / (n**2 * (n - 1))) + (y_std**4 / (m**2 * (m - 1))))
t_critical_welch = t.ppf(1 - alpha, df=df_welch)

# 进行假设检验
if t_statistic_welch < -t_critical_welch:
    print("(c) t_statistic_welch is:"+ str(t_statistic_welch)+"  t_critical_welch is:" + str(-t_critical_welch))
    print("So reject H0: μX > μY")
else:
    print("(c) t_statistic_welch is:"+ str(t_statistic_welch))
    print("Fail to reject H0")

# (d) 进行 F-Test
# H0: σ2X = σ2Y, H1: σ2X ≠ σ2Y
f_statistic = (x_std**2) / (y_std**2)

# 计算 F 临界值
f_critical_low = stats.f.ppf(alpha / 2, dfn=n - 1, dfd=m - 1)
f_critical_high = stats.f.ppf(1 - alpha / 2, dfn=n - 1, dfd=m - 1)

# 进行假设检验
if f_statistic < f_critical_low or f_statistic > f_critical_high:
    print("(d) f_statistic is:"+ str(f_statistic)+" f_critical_low is:"+"f_critical_low"+" f_critical_high is:"+"f_critical_high")
    print("(d) Reject H0: σ2X ≠ σ2Y")
else:
    print("(d) f_statistic is:"+ str(f_statistic)+" f_critical_low is:"+str(f_critical_low)+" f_critical_high is:"+str(f_critical_high))
    print("Fail to reject H0")
```
